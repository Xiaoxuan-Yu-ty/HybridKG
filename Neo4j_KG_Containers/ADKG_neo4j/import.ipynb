```python
import pandas as pd
import numpy as np
import networkx as nx
import pickle
```

```python
kg_path = "../../data/KG/ad_network_with_reverse_edges.pkl"

with open(kg_path, 'rb') as f:
    KG = pickle.load(f)
```

# Export networkx Graph to Neo4j compatible nodes.csv and edges.scv
+ in nodes.csv:
  + Neo4j requires the unique identifier column to be named :ID
  + the label column to be named :LABEL
+ in edges.csv
  + Neo4j requires :START_ID and :END_ID;
  + the edge type to be named :TYPE

```python
with open(kg_path, 'rb') as f:
    KG = pickle.load(f)

# 1. Convert Nodes
nodes_data = []
for node_id, attrs in KG.nodes(data=True):
    #print(attrs)
    node_dict = {":ID": node_id}
    
    if "label" in attrs:
        node_dict[":LABEL"] = attrs.pop("label")
    
    # Add all remaining attributes as Neo4j properties
    node_dict.update(attrs)
    nodes_data.append(node_dict)

df_nodes = pd.DataFrame(nodes_data)
df_nodes.to_csv("./import/nodes.csv", index=False)
df_nodes
```

```python
# 2. extract and format edges
edges_data = []
for source, target, rel, attrs in KG.edges(data=True, keys=True):
    # print(source, rel, target)
    # print(attrs)
    # break
    edge_dict = {
        ":START_ID": source,
        ":END_ID": target,
        ":TYPE": rel,
    }
         
    # Add all remaining attributes as edge properties
    edge_dict.update(attrs)
    edges_data.append(edge_dict)

df_edges = pd.DataFrame(edges_data)
df_edges.to_csv("./import/edges.csv", index=False)

print("Export complete! Files saved as 'nodes.csv' and 'edges.csv'.")
df_edges
```
